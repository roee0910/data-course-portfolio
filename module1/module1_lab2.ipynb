## Hands-On Problem 1.1: Which feed is best for business?

**Goal:** Compare minimum, mean, and maximum chick weight for each feed type, and recommend which feed to buy — defending the choice against at least one alternative.

**Repo URL:** _(https://github.com/roee0910/data-course-portfolio.git)_

## 1. Load the data

```python
import pandas as pd

df = pd.read_csv('chickwts.csv', index_col=0)
df.head()
```

## 2. Per-feed summary table

For each feed type, compute the minimum, mean, and maximum weight.

```python
summary = df.groupby('feed')['weight'].agg(['min', 'mean', 'max']).round(1)
summary = summary.sort_values('mean', ascending=False)
summary
```

## 3. Quick visual check (optional but helpful)

A bar chart of the mean weight per feed, so the differences are easy to see at a glance.

```python
summary['mean'].plot(kind='bar', title='Mean chick weight by feed type', ylabel='Weight (g)')
```
